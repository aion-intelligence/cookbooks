Boltz-2 is a state-of-the-art protein structure prediction model. It leverages deep learning techniques to accurately predict the 3D structure of proteins from their amino acid sequences.

In this notebook, we will demonstrate how to use Boltz-2 to predict protein structures and visualisation them using Mol*.

```python
import logging
from pathlib import Path

from utils.gpu import check_gpu
from utils.ui import create_input_ui, display_ui
from utils.boltz.prediction_and_viz import run_prediction_and_viz, show_structure_and_confidence
from utils.input_sequences import INSULIN_YAML

logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

MODELS_DIR = Path("./runs/boltz/models")
OUTPUT_DIR = Path("./runs/boltz/outputs")

# load example structure
cif_file = "./examples/insulin.cif"
confidence_file = "./examples/insulin_confidence.json"

_ = check_gpu()  # Check GPU availability
```

# Build and display the input UI

```python
# create UI components and attach prediction to Run button
ui_components = create_input_ui()

prediction_widget = ui_components["prediction_output"]
with prediction_widget:
    show_structure_and_confidence(cif_file, confidence_file)

# create prediction function that can be attached to run button
run_prediction_args = (MODELS_DIR, OUTPUT_DIR, ui_components)
run_prediction_fn = lambda b: run_prediction_and_viz(b, *run_prediction_args)

ui_components["run_button"].on_click(run_prediction_fn)
```

```python
# Display UI
display_ui("Boltz-2", ui_components)
```
