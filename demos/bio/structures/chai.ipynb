Chai-1 is an advanced protein structure prediction model developed by Chai Lab, designed to accurately determine the three-dimensional structures of proteins from their amino acid sequences. Building upon the success of models like AlphaFold, Chai-1 leverages deep learning techniques and extensive training on known protein structures to deliver high-precision predictions.

In this demonstration, we showcase how to utilize Chai-1 with GPU acceleration to efficiently predict protein structures, enabling researchers to gain insights into protein function and interactions.

```python
import logging
from pathlib import Path

from utils.gpu import check_gpu
from utils.ui import create_input_ui, display_ui
from utils.chai.prediction_and_viz import run_prediction_and_viz

logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

DEMO_ONLY = False  # Set to False to enable new predictions

MODELS_DIR = Path("./runs/chai/models")
OUTPUT_DIR = Path("./runs/chai/outputs")

if not check_gpu():  # Check GPU availability
    logger.info("Chai-1 require CPU to run. Running new predictions will fail")
```

# Build Chai UI

```python
# create UI components
ui_components = create_input_ui()

# create prediction function that to execute on button click
run_prediction_args = (MODELS_DIR, OUTPUT_DIR, ui_components)
run_prediction_fn = lambda b: run_prediction_and_viz(b, *run_prediction_args)

# attach prediction function to run button
ui_components["run_button"].on_click(run_prediction_fn)
```

```python
# Display UI
display_ui("Chai", ui_components, demo=DEMO_ONLY)
```
